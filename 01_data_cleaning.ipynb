```python
# take first column text only
gw_text = gw_raw.iloc[:,0].astype(str)

# keep rows containing comma-separated district data
gw_text = gw_text[gw_text.str.contains(",")]

# split by commas
gw_split = gw_text.str.split(",", expand=True)

# keep district + first groundwater level
gw_clean = gw_split[[0,1]].copy()
gw_clean.columns = ["District","GW_Level"]

# convert numeric
gw_clean["GW_Level"] = pd.to_numeric(gw_clean["GW_Level"], errors="coerce")

# remove junk rows
gw_clean = gw_clean.dropna(subset=["GW_Level"])

print(gw_clean.head(20))
print(gw_clean.shape)
```

```python
master = master.fillna(master.mean(numeric_only=True))
```

```python
master["Risk"] = "Low"

master.loc[
    (master["GW_Level"] < 5) |
    (master["Electric Conductivity (µS/cm)"] > 1500),
    "Risk"
] = "Medium"

master.loc[
    (master["GW_Level"] < 3) |
    (master["Electric Conductivity (µS/cm)"] > 3000),
    "Risk"
] = "High"

print(master[["District","Risk"]].head(20))
```

```python
master["Risk"] = "Low"

master.loc[
    (master["GW_Level"] < 5) |
    (master["Electric Conductivity (μS/cm)"] > 1500),
    "Risk"
] = "Medium"

master.loc[
    (master["GW_Level"] < 3) |
    (master["Electric Conductivity (μS/cm)"] > 3000),
    "Risk"
] = "High"

print(master[["District","Risk"]].head(20))
```

```python
X = master.select_dtypes(include=["float64","int64"]).copy()
y = master["Risk"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report
```

```python
le = LabelEncoder()
y_encoded = le.fit_transform(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.25, random_state=42
)
```

```python
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)
```

```python
pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
```

```python
master["Predicted_Risk"] = le.inverse_transform(model.predict(X))

print(master[["District","Risk","Predicted_Risk"]].head(20))
```

```python
master.to_csv("final_groundwater_results.csv", index=False)
```
